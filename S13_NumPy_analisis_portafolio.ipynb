## 00 · ¿Qué es una librería?

> **Nota** — Objetivo: entender qué es una librería antes de usar una.

Una librería es un paquete de funciones que otra persona ya escribió, probó y publicó para que cualquiera las use — como las tuyas, pero construidas para casos generales. **NumPy** es una librería con funciones ya hechas para trabajar con muchos números a la vez.

**¿Cuál es el objetivo de NumPy?** Darle a Python la capacidad de hacer cálculos numéricos rápido sobre conjuntos grandes de datos.

````python
import numpy as np

print("NumPy", np.__version__, "listo para usar.")
````

> 💡 **Idea clave** — `import` le dice a Python que traiga la librería. `numpy` es su nombre. `as np` le da un apodo corto — es la convención que usa toda la comunidad.

## 01 · Qué es un array — y cuántas dimensiones puede tener

> **Nota** — Objetivo: entender qué es un array, y que puede tener una o más dimensiones.

Un array es una colección de números, todos del mismo tipo, donde cada uno tiene una posición fija — su **índice**.

````python
precios = np.array([195, 385, 185, 165])   # AAPL, MSFT, AMZN, GOOGL

print(precios)
print("shape:", precios.shape)
print("dtype:", precios.dtype)
````

Esta fila tiene **una sola dimensión** — un **eje (axis)**. Pero un array puede tener dos ejes: una tabla, con filas y columnas. Por ejemplo, el precio de 4 activos durante 5 días:

| | Día 1 | Día 2 | Día 3 | Día 4 | Día 5 |
|---|---|---|---|---|---|
| **AAPL** | 192 | 194 | 193 | 196 | 195 |
| **MSFT** | 390 | 388 | 386 | 384 | 385 |

````python
precios_5_dias = np.array([
    [192, 194, 193, 196, 195],   # AAPL
    [390, 388, 386, 384, 385],   # MSFT
])

print(precios_5_dias)
print("shape:", precios_5_dias.shape, "  # (2 activos, 5 días)")
````

> 💡 **Idea clave** — Más adelante en esta clase, cada activo va a tener justo esta forma: **una fila de rendimientos a través de varios días**. Guarda esta idea — es la base de todo lo que sigue.

## 03 · Operaciones básicas

> **Nota** — Objetivo: operar el array completo, sin ciclos.

Los operadores aritméticos se aplican **elemento por elemento**.

````python
precios_actuales = np.array([195, 385, 185, 165])
precios_compra = np.array([180, 400, 170, 150])

valor_invertido = cantidades * precios_compra
valor_actual = cantidades * precios_actuales

print("Valor actual:", valor_actual)
````

**✏️ Completa el código** — calcula `ganancia` y `en_perdida` (verdadero/falso).

````python
ganancia = valor_actual - valor_invertido
en_perdida = ganancia < 0

print("Ganancia:", ganancia)
print("En pérdida:", en_perdida)
````

## 04 · Analiza un activo: rendimiento y riesgo en el tiempo

> **Nota** — Objetivo: usar `mean()` y `std()` para describir el comportamiento de un activo a través de varios días — no solo calcularlos.

A partir de aquí trabajas con algo distinto: no es "un número por activo", sino **la serie de rendimientos diarios de AAPL durante 8 días** de negociación.

````python
AAPL = np.array([0.012, 0.008, -0.004, 0.015, 0.006, -0.002, 0.010, 0.003])

print("Número de días:", len(AAPL))
print("Primer día:", AAPL[0])
print("Último día:", AAPL[-1])
````

¿Tiene razón? Usa el propio array `AAPL` (no solo el promedio) para comprobarlo antes de responder.

````python
promedio_aapl = AAPL.mean()
peor_dia = AAPL.min()

print("Promedio:", promedio_aapl)
print("Peor día:", peor_dia)
print("Días negativos:", (AAPL < 0).sum())
````

El promedio resume la tendencia de los 8 días; no dice que cada día individual haya sido bueno. Para afirmar "todos los días fueron buenos" hay que revisar la serie completa, no solo su resumen.

## 05 · Compara los 4 activos

> **Nota** — Objetivo: comparar rendimiento y riesgo de varios activos a la vez, y tomar una posición justificada — no solo calcular cada uno por separado.

Estas son las series de 8 días de tus 4 activos:

````python
AAPL = np.array([0.012, 0.008, -0.004, 0.015, 0.006, -0.002, 0.010, 0.003])
MSFT = np.array([0.004, 0.003, 0.005, 0.002, 0.006, 0.004, 0.003, 0.005])
AMZN = np.array([-0.010, -0.015, 0.002, -0.008, -0.012, 0.005, -0.020, -0.006])
TSLA = np.array([0.030, -0.025, 0.040, 0.015, -0.020, 0.035, -0.010, 0.020])
````

**✏️ Completa la tabla** — calcula el rendimiento promedio y el riesgo de cada uno.

````python
activos = {"AAPL": AAPL, "MSFT": MSFT, "AMZN": AMZN, "TSLA": TSLA}

for nombre, serie in activos.items():
    promedio = serie.mean()
    riesgo = serie.std()
    print(f"{nombre}: rendimiento {promedio:.4f}  |  riesgo {riesgo:.4f}")
````

## 06 · Construye un portafolio ponderado

> **Nota** — Objetivo: combinar varios activos con distintos pesos, y entender qué le pasa al riesgo cuando diversificas.

Un portafolio real casi nunca es "todo en un solo activo" — se reparte el dinero entre varios, con distintos pesos. Supón esta distribución:

````python
pesos = np.array([0.35, 0.40, 0.10, 0.15])  # AAPL, MSFT, AMZN, TSLA — suman 1.0

rendimiento_portafolio = AAPL*pesos[0] + MSFT*pesos[1] + AMZN*pesos[2] + TSLA*pesos[3]
print(rendimiento_portafolio)
````

> 💡 **Idea clave** — Una sola operación combinó los 4 activos, día por día, sin ningún `for`. Cada día del portafolio es el promedio ponderado de lo que hicieron los 4 activos ese día.

**✏️ Completa el código** — calcula el rendimiento promedio y el riesgo del portafolio.

````python
rendimiento_promedio_portafolio = rendimiento_portafolio.mean()
riesgo_portafolio = rendimiento_portafolio.std()

print(f"Rendimiento promedio del portafolio: {rendimiento_promedio_portafolio:.4f}")
print(f"Riesgo del portafolio: {riesgo_portafolio:.4f}")
````

<details>
<summary>💡 Ver respuesta</summary>

```python
rendimiento_promedio_portafolio = rendimiento_portafolio.mean()
riesgo_portafolio = rendimiento_portafolio.std()
```

Rendimiento promedio ≈ **0.45%**, riesgo ≈ **0.36%**.
</details>

**🔎 Pregunta de análisis**

Compara el riesgo del portafolio (≈0.36%) con el promedio ponderado de los riesgos *individuales* de los 4 activos, calculado abajo:

````python
riesgos_individuales = np.array([AAPL.std(), MSFT.std(), AMZN.std(), TSLA.std()])
riesgo_promedio_simple = (riesgos_individuales * pesos).sum()

print(f"Riesgo si solo promediaras los riesgos individuales: {riesgo_promedio_simple:.4f}")
print(f"Riesgo real del portafolio combinado: {riesgo_portafolio:.4f}")
````

El riesgo real del portafolio (≈0.36%) es **casi la mitad** de lo que esperarías si solo promediaras el riesgo de cada activo por separado (≈0.70%). Ningún activo individual cambió de riesgo — el portafolio completo sí.

**Explica con tus palabras**, en una celda de texto, por qué combinar activos distintos puede reducir el riesgo total sin que ninguno de ellos se vuelva menos riesgoso por su cuenta. (Pista: fíjate en qué días TSLA pierde y qué hacen AAPL o MSFT esos mismos días.)

````python
# Evidencia: ¿qué hacen AAPL y MSFT los días en que TSLA pierde?
dias_tsla_pierde = TSLA < 0
print("TSLA pierde los días:", np.where(dias_tsla_pierde)[0] + 1)
print("AAPL esos días:", AAPL[dias_tsla_pierde])
print("MSFT esos días:", MSFT[dias_tsla_pierde])
print("Correlación TSLA-AAPL:", np.corrcoef(TSLA, AAPL)[0, 1].round(2))
````

**✍️ Mi explicación (sección 06)**

Combinar activos reduce el riesgo porque no se mueven igual los mismos días. TSLA pierde los días 2, 5 y 7, y justo esos días AAPL gana (+0.8 %, +0.6 % y +1.0 %) y MSFT también (+0.3 %, +0.6 % y +0.3 %), así que sus ganancias amortiguan la caída de TSLA. La correlación entre TSLA y AAPL es negativa (≈ -0.43): tienden a moverse en direcciones opuestas. Por eso el riesgo del portafolio (≈ 0.36 %) es casi la mitad del promedio ponderado de los riesgos individuales (≈ 0.70 %), aunque ningún activo haya cambiado su propio riesgo: ese efecto es la **diversificación**.


<details>
<summary>💡 Ver una forma de responder</summary>

Porque los activos no suben y bajan exactamente el mismo día ni en la misma dirección. Cuando uno pierde, otro puede sostenerse o ganar, y eso amortigua el resultado combinado — es el efecto de la **diversificación**.

Dicho de otra forma: el riesgo de un portafolio no depende solo de cuánto se mueve cada activo por separado, sino de qué tan parecido se mueven entre sí — su **correlación**. Cuando dos activos tienden a moverse en direcciones opuestas (uno sube cuando el otro baja), combinarlos reduce el riesgo total mucho más que si simplemente promediaras sus riesgos individuales. Es la razón por la que un portafolio bien armado casi siempre mezcla activos que no reaccionan igual a las mismas condiciones del mercado.
</details>

## 🎯 07 · Reto individual

Retoma el portafolio que te asignaron en el reto de S11-S12.

````python
# Datos asignados a mí (los de "Copiar mis datos para el cuaderno").
# Misma estructura que el diccionario de la plantilla: cantidad y precio_compra por activo.
mi_portafolio = {
    "NFLX": {"cantidad": 16, "precio_compra": 284},
    "V":    {"cantidad": 14, "precio_compra": 295},
    "TSLA": {"cantidad": 15, "precio_compra": 423},
    "JPM":  {"cantidad": 8,  "precio_compra": 288},
}
mis_precios = {"NFLX": 298, "V": 306, "TSLA": 352, "JPM": 232}

print("Mi portafolio asignado:")
print(mi_portafolio)
print("Precios actuales:", mis_precios)
````

### Tu tarea

Con `mi_portafolio` y `mis_precios`, usando **solo arrays de NumPy**:

1. Calcula `valor_invertido`, `valor_actual`, `ganancia` y `rendimiento` (%) de tus 4 activos.
2. Calcula el promedio, el mejor, el peor y la desviación estándar del rendimiento.
3. **Escribe una recomendación de una sola frase** para tu portafolio, dirigida a un inversionista, que considere tanto el rendimiento como el riesgo — no solo uno de los dos. Debe quedar claro, leyendo tu frase, en qué números te basaste.

````python
import numpy as np

# --- 1) Arrays (mismo orden de activos en los tres) ---
activos = list(mi_portafolio.keys())
cantidades       = np.array([mi_portafolio[a]["cantidad"] for a in activos])
precios_compra   = np.array([mi_portafolio[a]["precio_compra"] for a in activos])
precios_actuales = np.array([mis_precios[a] for a in activos])

valor_invertido = cantidades * precios_compra
valor_actual    = cantidades * precios_actuales
ganancia        = valor_actual - valor_invertido
rendimiento     = ganancia / valor_invertido * 100   # en %

print(f"{'Activo':<6}{'Invertido':>10}{'Actual':>9}{'Ganancia':>10}{'Rend. %':>9}")
for a, vi, va, g, r in zip(activos, valor_invertido, valor_actual, ganancia, rendimiento):
    print(f"{a:<6}{vi:>10}{va:>9}{g:>10}{r:>9.2f}")

# --- 2) Estadísticas del rendimiento ---
promedio     = rendimiento.mean()
mejor        = rendimiento.max()
peor         = rendimiento.min()
riesgo       = rendimiento.std()          # desviación estándar (ddof=0, por defecto en NumPy)
mejor_activo = activos[rendimiento.argmax()]
peor_activo  = activos[rendimiento.argmin()]

print(f"\nPromedio: {promedio:.2f} %")
print(f"Mejor:    {mejor:.2f} % ({mejor_activo})")
print(f"Peor:     {peor:.2f} % ({peor_activo})")
print(f"Desviación estándar: {riesgo:.2f}")

# Dato adicional: rendimiento total del portafolio ponderado por lo invertido
rend_total = ganancia.sum() / valor_invertido.sum() * 100
print(f"\nRendimiento total ponderado: {rend_total:.2f} %  (ganancia total: {ganancia.sum()})")

# --- 3) Recomendación (una sola frase) ---
perdedores = " y ".join(a for a, r in zip(activos, rendimiento) if r < 0)
recomendacion = (
    f"Recomiendo no aumentar la inversión y reevaluar las posiciones en {perdedores}, "
    f"porque el rendimiento promedio es {promedio:.2f} % con una desviación estándar de {riesgo:.2f}: "
    f"un resultado negativo y muy disperso, con un rango de {peor:.2f} % a {mejor:.2f} %."
)
print("\n" + recomendacion)
````
