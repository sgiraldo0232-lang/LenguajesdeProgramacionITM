## 00 · ¿Qué es pandas y para qué se usa?

> **Nota** — Objetivo: entender qué problema resuelve pandas, y por qué no basta con lo que ya sabes de NumPy.

Un portafolio real no es solo números: tiene tickers (texto), cantidades (enteros), precios (decimales) y, más adelante, fechas. NumPy asume que todo el array es del mismo tipo — por eso necesitas algo que junte columnas de distinto tipo en una sola tabla, con nombres. Eso es exactamente lo que resuelve **pandas**.

pandas es una librería para trabajar con datos organizados en tablas: leerlos, explorarlos, limpiarlos, filtrarlos y resumirlos. Está construida sobre NumPy — por dentro, cada columna de una tabla de pandas es un array.

> 💡 **Idea clave** — NumPy responde *"¿cómo calculo rápido sobre muchos números?"*. pandas responde *"¿cómo organizo datos de distinto tipo en algo que se pueda leer, filtrar y consultar como una tabla?"*. No compiten — pandas usa NumPy por dentro.

📚 Fuente: ["Package overview", pandas.pydata.org](https://pandas.pydata.org/docs/getting_started/overview.html)

## 01 · La estructura: Series y DataFrame

> **Nota** — Objetivo: distinguir las dos estructuras de datos que ofrece pandas.

pandas ofrece dos tipos de estructuras:

- **Series** — un arreglo unidimensional etiquetado, capaz de contener datos de cualquier tipo. Es una sola columna, con índice.
- **DataFrame** — una estructura de dos dimensiones que guarda datos como una tabla, con filas y columnas. Cada columna de un DataFrame es, de hecho, una Series.

````python
import numpy as np
import pandas as pd

precios = pd.Series([195, 385, 185, 165])
print(precios)
print(type(precios))
````

````python
portafolio = pd.DataFrame({
    "ticker": ["AAPL", "MSFT", "AMZN", "GOOGL"],
    "cantidad": [10, 8, 12, 6],
    "precio": [195, 385, 185, 165]
})
print(portafolio)
print(type(portafolio))
````

## 05 · Operaciones más usadas

> **Nota** — Objetivo: agregar columnas calculadas y resumir un DataFrame sin ciclos.

Agregar una columna nueva es tan simple como asignarla — pandas alinea los valores automáticamente.

**✏️ Completa el código**

````python
portafolio["valor_actual"] = portafolio["cantidad"] * portafolio["precio_actual"]
portafolio["ganancia"] = portafolio["valor_actual"] - (portafolio["cantidad"] * portafolio["precio_compra"])

print(portafolio["ganancia"].mean())   # promedio
print(portafolio["ganancia"].std())   # desviación estándar
portafolio.sort_values(by="ganancia")
````

## 06 · Todo junto: el portafolio como DataFrame

> **Nota** — Objetivo: integrar creación, columnas calculadas, filtro y orden en un solo flujo.

````python
portafolio = pd.DataFrame({
    "ticker": ["AAPL", "MSFT", "AMZN", "GOOGL"],
    "cantidad": [10, 8, 12, 6],
    "precio_compra": [180, 400, 170, 150],
    "precio_actual": [195, 385, 185, 165]
})

portafolio["valor_actual"] = portafolio["cantidad"] * portafolio["precio_actual"]
portafolio["ganancia"] = portafolio["valor_actual"] - (portafolio["cantidad"] * portafolio["precio_compra"])
portafolio["rendimiento"] = (portafolio["ganancia"] / (portafolio["cantidad"] * portafolio["precio_compra"])) * 100

en_perdida = portafolio[portafolio["rendimiento"] < 0]
portafolio.sort_values(by="rendimiento", ascending=False)
````

**🔎 Pregunta de análisis**

Compara este bloque de código con el que escribiste en S11-S12 usando `for` y diccionarios. ¿Qué parte del código viejo desapareció por completo? ¿Qué se volvió más corto y por qué?

**✍️ Mi respuesta**

**¿Qué desapareció por completo?** El ciclo `for` que recorría el diccionario activo por activo, junto con las variables que se iban llenando dentro del ciclo (acumuladores, listas auxiliares) y el acceso manual a cada valor por su clave. En pandas no hay que decirle a Python *cómo* recorrer los datos: se le dice *qué* calcular.

**¿Qué se volvió más corto y por qué?** Los tres cálculos (`valor_actual`, `ganancia`, `rendimiento`) pasaron de ser operaciones repetidas dentro de un ciclo a una sola línea por columna, porque pandas aplica la operación a toda la columna a la vez (vectorizado, sobre NumPy). El filtro de los activos en pérdida pasó de un `if` dentro del ciclo con un `append` a una sola condición (`portafolio["rendimiento"] < 0`), y el orden pasó de ordenar listas o diccionarios a mano a `sort_values`. El resultado es menos código, más fácil de leer y con menos lugares donde equivocarse.

## 🎯 07 · Reto individual

Retoma el portafolio que te asignaron en el reto de S11-S12.

````python
# Datos asignados a mí (los de "Copiar mis datos para el cuaderno"; son los mismos de S11-S12 y S13).
# Misma estructura que la plantilla: cantidad y precio_compra por activo, y precios de hoy aparte.
mi_portafolio_dict = {
    "NFLX": {"cantidad": 16, "precio_compra": 284},
    "V":    {"cantidad": 14, "precio_compra": 295},
    "TSLA": {"cantidad": 15, "precio_compra": 423},
    "JPM":  {"cantidad": 8,  "precio_compra": 288},
}
mis_precios_dict = {"NFLX": 298, "V": 306, "TSLA": 352, "JPM": 232}

print("Mi portafolio asignado:")
print(mi_portafolio_dict)
print("Precios actuales:", mis_precios_dict)
````

### Tu tarea

Con `mi_portafolio_dict` y `mis_precios_dict`:

1. Construye un DataFrame con las columnas `ticker`, `cantidad`, `precio_compra`, `precio_actual`.
2. Agrega las columnas `valor_actual`, `ganancia` y `rendimiento`.
3. Filtra los activos en pérdida.
4. Ordena el DataFrame de mayor a menor rendimiento.

````python
import pandas as pd
from IPython.display import display

# --- 1) DataFrame: ticker, cantidad, precio_compra, precio_actual ---
mi_df = (
    pd.DataFrame.from_dict(mi_portafolio_dict, orient="index")   # una fila por activo
      .rename_axis("ticker")
      .reset_index()                                              # ticker pasa a ser columna
)
mi_df["precio_actual"] = mi_df["ticker"].map(mis_precios_dict)    # alinea el precio de hoy por ticker

print("1) DataFrame base")
display(mi_df)

# --- 2) Columnas calculadas ---
mi_df["valor_actual"] = mi_df["cantidad"] * mi_df["precio_actual"]
mi_df["ganancia"] = mi_df["valor_actual"] - (mi_df["cantidad"] * mi_df["precio_compra"])
mi_df["rendimiento"] = (mi_df["ganancia"] / (mi_df["cantidad"] * mi_df["precio_compra"])) * 100

print("\n2) Con valor_actual, ganancia y rendimiento (%)")
display(mi_df.round(2))

# --- 3) Activos en pérdida ---
en_perdida = mi_df[mi_df["rendimiento"] < 0]

print("\n3) Activos en pérdida")
display(en_perdida.round(2))

# --- 4) Orden de mayor a menor rendimiento ---
ordenado = mi_df.sort_values(by="rendimiento", ascending=False)

print("\n4) Ordenado de mayor a menor rendimiento")
display(ordenado.round(2))

# --- Datos para el exit ticket ---
primero = ordenado.iloc[0]["ticker"]
ultimo = ordenado.iloc[-1]["ticker"]
n_perdida = len(en_perdida)
print(f"\nPrimero: {primero} | Último: {ultimo} | Activos en pérdida: {n_perdida}")
````
